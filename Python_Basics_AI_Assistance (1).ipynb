```python
def factorial_iterative(n):
    if n < 0:
        return "Factorial is not defined for negative numbers."
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

def factorial_recursive(n):
    if n < 0:
        return "Factorial is not defined for negative numbers."
    if n == 0 or n == 1:
        return 1
    return n * factorial_recursive(n - 1)

# Example usage:
num = 5
print(f"Factorial of {num} (Iterative): {factorial_iterative(num)}")
print(f"Factorial of {num} (Recursive): {factorial_recursive(num)}")
```

```python
def generate_fibonacci(n):
    if n <= 0:
        return []
    elif n == 1:
        return [0]

    sequence = [0, 1]
    while len(sequence) < n:
        sequence.append(sequence[-1] + sequence[-2])
    return sequence

# Example usage:
n_terms = 10
fib_sequence = generate_fibonacci(n_terms)
print(f"Fibonacci sequence up to {n_terms} terms: {fib_sequence}")
```

```python
import re

def c_struct_to_dict(c_struct_str):
    # Extract the block inside the curly braces
    match = re.search(r'struct\s+\w+\s*\{(.*?)\}', c_struct_str, re.DOTALL)
    if not match:
        return "Invalid C struct format."

    struct_body = match.group(1)

    # Split body by semicolons
    lines = struct_body.split(';')
    struct_dict = {}

    for line in lines:
        line = line.strip()
        if not line:
            continue

        # Regex to match type and variable name (e.g., 'int age' or 'char name[50]')
        pattern = r'(?P<type>[\w\s\*]+?)\s+(?P<name>\w+)(?P<array>\[\d*\])?'
        var_match = re.search(pattern, line)

        if var_match:
            var_type = var_match.group('type').strip()
            var_name = var_match.group('name').strip()
            is_array = var_match.group('array') is not None

            # Map C types to Pythonic descriptions or defaults
            struct_dict[var_name] = {
                "type": f"{var_type}[]" if is_array else var_type
            }

    return struct_dict

# Example C Struct definition
c_struct = """
struct Person {
    char name[100];
    int age;
    float height;
    double *coordinates;
};
"""

resulting_dict = c_struct_to_dict(c_struct)
import json
print(json.dumps(resulting_dict, indent=4))
```
